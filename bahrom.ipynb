```python
from docx import Document

doc=Document()

doc.add_heading('mening hujatm',level=1)

doc.add_heading('kirish qismi',level=1)

doc.add_paragraph('Bu paragraph odiy matn.istalgan uzunlikda yozsa bo\'lado.')

doc.save('ksdffhkj.docx')
print('Fayl yaratldi: birinchi_hujjat.docx')
```

```python
from docx import Document
from docx.shared import Pt, RGBColor

doc = Document()
doc.add_heading('Shrift namunalari', level=1)


p = doc.add_paragraph()

r1 = p.add_run('Oddiy matn.  ')

r2 = p.add_run('Bu qalin matn.  ')
r2.bold = True
r3 = p.add_run('Bu kursiv matn.  ')
r3.italic = True


r4 = p.add_run('Tagiga chizilgan.')
r4.underline = True


p2 = doc.add_paragraph()

r5 = p2.add_run('Bu 18pt qizil matn!')
r5.font.size = Pt(18)
r5.font.color.rgb = RGBColor(0xFF, 0x00, 0x00)

r6 = p2.add_run('  Bu 14pt ko\'k matn!')
r6.font.size = Pt(14)
r6.font.color.rgb = RGBColor(0x00, 0x70, 0xC0)

doc.save('word_shrift.docx')
print('word_shrift.docx saqlandi')
```

```python
from openpyxl import Workbook


wb = Workbook()
ws = wb.active
ws.title = "O'quvchilar"


ws['A1'] = '№'
ws['B1'] = 'Ism'
ws['C1'] = 'Familiya'
ws['D1'] = 'Yoshi'
ws['E1'] = 'Bahosi'


malumotlar = [
    (1, 'Ali', 'Valiyev', 15, 5),
    (2, 'Zulfiya', 'Rahimova', 16, 4),
    (3, 'Bobur', 'Toshmatov', 15, 5),
    (4, 'Malika', 'Yusupova', 17, 4),
]

for i, qator in enumerate(malumotlar, start=2):
    for j, qiymat in enumerate(qator, start=1):
        ws.cell(row=i, column=j, value=qiymat)

wb.save('excel_birinchi.xlsx')
print('✅ excel_birinchi.xlsx saqlandi')
```
